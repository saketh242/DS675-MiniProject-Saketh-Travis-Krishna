```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
df_data = pd.read_csv('data/Training_indian_liver_disease_dataset.csv')

df_features = df_data[['Age',
                       'BMI',
                       'Sleep_Hours',
                       'ALT',
                       'AST',
                       'Bilirubin',
                       'Albumin',
                       'Platelets',
                       'Alk_Phosphatase'
                       ]].copy()

df_features['Gender'] = (df_data['Gender'] == 'Male').astype(int)
df_features['Liver_Disease_Type'] = (df_data['Liver_Disease_Type'] == 'Normal').astype(int)
X_train = df_features.iloc[:, 1:].values
y_train = df_features.iloc[:, -1].values

scaler = MinMaxScaler(feature_range=(0, 1))
X_train = scaler.fit_transform(X_train)

print(X_train)
print(y_train)
```
